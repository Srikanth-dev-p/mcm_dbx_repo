```python
%sql
CREATE TABLE db_cms_v1.default.patient (
ID BIGINT GENERATED ALWAYS AS IDENTITY NOT NULL,
 
ProviderID INT,
MBI STRING NOT NULL,
HICN STRING,
 
FirstName STRING NOT NULL,
MiddleName STRING,
LastName STRING NOT NULL,
 
DOB TIMESTAMP NOT NULL,
Gender STRING NOT NULL,
RaceCode STRING,
DOD TIMESTAMP,
 
AddrLine1 STRING,
AddrLine2 STRING,
City STRING,
State STRING,
Zip STRING,
 
Phone STRING,
Cell STRING,
OtherPhone STRING,
Email STRING,
 
FIPSState STRING,
FIPSCounty STRING,
EnrollmentZip STRING,
 
MedicareStatusCode STRING,
DualStatusCode STRING,
EntitlementReasonCode STRING,
EntitlementBuyInCode STRING,
 
PartAEnrollDate TIMESTAMP,
PartBEnrollDate TIMESTAMP,
 
HospiceEnrollDateStart TIMESTAMP,
HospiceEnrollDateEnd TIMESTAMP,
 
SSAStateCounty STRING,
FIPSStateCounty STRING,
 
AlignmentStartDate TIMESTAMP,
AlignmentEndDate TIMESTAMP,
 
PreferredLanguage STRING,
 
ProcessLogID INT NOT NULL,
 
MIPSCQMInd BOOLEAN,
IsActive BOOLEAN,
 
DisplayName STRING NOT NULL,
HIEMemberID STRING,
 
UpdatedBy INT,
UpdatedDate TIMESTAMP,
 
CreatedBy INT NOT NULL,
CreatedDate TIMESTAMP,
 
ExternalMemberID STRING,
ContractPBP STRING,
 
CONSTRAINT pk_patient PRIMARY KEY (ID)
)
USING DELTA;
```
